```python
import sqlite3
```

```python
# Membuat koneksi database
conn = sqlite3.connect('perpustakaan.db')

# Membuat cursor
cursor = conn.cursor()

print("Database berhasil dibuat!")
```

```python
cursor.execute('''
CREATE TABLE buku (
    id INTEGER PRIMARY KEY,
    judul TEXT,
    penulis TEXT,
    tahun INTEGER
)
''')

print("Tabel buku berhasil dibuat!")
```

Membuat Tabel Anggota

```python
cursor.execute('''
CREATE TABLE anggota (
    id INTEGER PRIMARY KEY,
    nama TEXT,
    email TEXT
)
''')

print("Tabel anggota berhasil dibuat!")
```

Membuat Tabel Peminjaman

```python
cursor.execute('''
CREATE TABLE peminjaman (
    id INTEGER PRIMARY KEY,
    id_anggota INTEGER,
    id_buku INTEGER,
    tgl_pinjam TEXT,
    tgl_kembali TEXT,

    FOREIGN KEY (id_anggota) REFERENCES anggota(id),
    FOREIGN KEY (id_buku) REFERENCES buku(id)
)
''')

print("Tabel peminjaman berhasil dibuat!")
```

A. Menambahkan 5 data buku

```python
data_buku = [
    (1, "Laskar Pelangi", "Andrea Hirata", 2005),
    (2, "Bumi", "Tere Liye", 2014),
    (3, "Negeri 5 Menara", "Ahmad Fuadi", 2009),
    (4, "Dilan 1990", "Pidi Baiq", 2014),
    (5, "Ayat-Ayat Cinta", "Habiburrahman El Shirazy", 2004)
]

cursor.executemany('''
INSERT INTO buku VALUES (?, ?, ?, ?)
''', data_buku)

print("Data buku berhasil ditambahkan!")
```

Menambahkan 3 data Anggota

```python
data_anggota = [
    (1, "Andi", "andi@gmail.com"),
    (2, "Siti", "siti@gmail.com"),
    (3, "Budi", "budi@gmail.com")
]

cursor.executemany('''
INSERT INTO anggota VALUES (?, ?, ?)
''', data_anggota)

print("Data anggota berhasil ditambahkan!")
```

B. Menambahkan 4 data transaksi peminjaman

```python
data_peminjaman = [
    (1, 1, 2, "2025-05-01", "2025-05-07"),
    (2, 2, 1, "2025-05-03", "2025-05-10"),
    (3, 3, 4, "2025-05-05", None),
    (4, 1, 5, "2025-05-08", "2025-05-15")
]

cursor.executemany('''
INSERT INTO peminjaman VALUES (?, ?, ?, ?, ?)
''', data_peminjaman)

print("Data peminjaman berhasil ditambahkan!")
```

Menyimpan perubahan database

```python
conn.commit()

print("Semua data berhasil disimpan!")
```

 C. Menampilkan daftar peminjaman lengkap (Join)

```python
query = '''
SELECT
    peminjaman.id,
    anggota.nama,
    buku.judul,
    peminjaman.tgl_pinjam,
    peminjaman.tgl_kembali
FROM peminjaman
JOIN anggota
    ON peminjaman.id_anggota = anggota.id
JOIN buku
    ON peminjaman.id_buku = buku.id
'''

cursor.execute(query)

hasil = cursor.fetchall()

print("Daftar Peminjaman:\n")

for row in hasil:
    print(row)
```

D. Menghitung Jumlah Buku yang Dipinjam Tiap Anggota (GROUP BY)

```python
query_group = '''
SELECT
    anggota.nama,
    COUNT(peminjaman.id_buku) AS jumlah_pinjaman
FROM peminjaman
JOIN anggota
    ON peminjaman.id_anggota = anggota.id
GROUP BY anggota.nama
'''

cursor.execute(query_group)

hasil_group = cursor.fetchall()

print("Jumlah Buku yang Dipinjam Setiap Anggota:\n")

for row in hasil_group:
    print(row)
```

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

A. Membuat DataFrame dari Hasil Query SQL

```python
query = '''
SELECT
    peminjaman.id,
    anggota.nama,
    buku.judul,
    peminjaman.tgl_pinjam,
    peminjaman.tgl_kembali
FROM peminjaman
JOIN anggota
    ON peminjaman.id_anggota = anggota.id
JOIN buku
    ON peminjaman.id_buku = buku.id
'''

df = pd.read_sql_query(query, conn)

print(df)
```

 Mengubah Kolom Tanggal Menjadi Format Datetime

```python
df['tgl_pinjam'] = pd.to_datetime(df['tgl_pinjam'])
df['tgl_kembali'] = pd.to_datetime(df['tgl_kembali'])
```

B. Menghitung Durasi Peminjaman

```python
df['durasi_hari'] = (df['tgl_kembali'] - df['tgl_pinjam']).dt.days

print(df)
```

C. Identifikasi Data yang Belum Dikembalikan

```python
belum_kembali = df[df['tgl_kembali'].isnull()]

print("Data buku yang belum dikembalikan:\n")
print(belum_kembali)
```

Menghitung Jumlah Buku yang Dipinjam Tiap Anggota

```python
jumlah_pinjaman = df.groupby('nama')['judul'].count()

print(jumlah_pinjaman)
```

Membuat Histogram Durasi Peminjaman

```python
plt.figure(figsize=(6,4))

plt.hist(df['durasi_hari'].dropna())

plt.title('Histogram Durasi Peminjaman')
plt.xlabel('Durasi (Hari)')
plt.ylabel('Jumlah Peminjaman')

plt.show()
```

Membuat Diagram Batang Jumlah Buku per Anggota

```python
jumlah_pinjaman.plot(kind='bar')

plt.title('Jumlah Buku yang Dipinjam per Anggota')
plt.xlabel('Nama Anggota')
plt.ylabel('Jumlah Buku')

plt.show()
```
